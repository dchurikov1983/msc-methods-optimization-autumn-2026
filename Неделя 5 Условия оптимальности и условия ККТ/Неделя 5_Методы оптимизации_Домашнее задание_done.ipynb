Решение задач на дом. Неделя 5

Задача 1 Проекция на коробочные ограничения

Коробочные ограничения $l_i \leqslant x_i \leqslant u_i$ появляются в проекции градиентного шага, если исходная задача содержит неотрицательные переменные и ограничения на мощности. Для заданных $\mathbf{y} \in \mathbb{R}^n$ и границ $l_i < u_i$ решите

$ \min_{\mathbf{x} \in \mathbb{R}^n} \frac{1}{2} \|\mathbf{x} - \mathbf{y}\|_2^2 \quad \text{при} \quad l_i \leqslant x_i \leqslant u_i, \quad i = 1, \dots, n. $


1.1 Выпишите условия ККТ. Множитель при xi −ui ⩽0 обозначьте μi,
множитель при li −xi ⩽0 обозначьте λi.


1. Сформулируем ограничения

Перепишем исходные коробочные ограничения $l_i \le x_i \le u_i$ в виде неравенств $f_i(x) \le 0$:

• Ограничения сверху: $x_i - u_i \le 0$ для всех $i = 1, \dots, n$. Множители Лагранжа: $\mu _{i}$.

• Ограничения снизу: $l_i - x_i \le 0$ для всех $i = 1, \dots, n$. Множители Лагранжа: $\lambda _{i}$.


2. Выпишем функцию Лагранжа (Лагранжиан)

Целевая функция: $f_0(x) = \frac{1}{2} \Vert{}x - y\Vert{}_2^2 = \frac{1}{2} \sum_{i=1}^n (x_i - y_i)^2$

из определения евклидовой нормы будет равна $\frac{1}{2} \sum_{i=1}^n (x_i - y_i)^2$

Полный лагранжиан будет иметь вид:
$L(x,\lambda ,\mu )=\frac{1}{2}\sum _{i=1}^{n}(x_{i}-y_{i})^{2}+\sum _{i=1}^{n}\mu _{i}(x_{i}-u_{i})+\sum _{i=1}^{n}\lambda _{i}(l_{i}-x_{i})$

3. условия ККТ

3.1 Стационарность

$\frac{\partial L}{\partial x_{i}}=x_{i}-y_{i}+\mu _{i}-\lambda _{i}=0\implies x_{i}-y_{i}+\mu _{i}-\lambda _{i}=0$

3.2 Допустимость прямой точки. Должны выполняться ограничения :

$x_{i}-u_{i}\le 0$

$l_{i}-x_{i}\le 0$


3.3 Допустимость двойственных переменных 

$\mu _{i}\ge 0$

$\lambda _{i}\ge 0$

3.4 Дополняющая нежёсткость 


$\mu _{i}(x_{i}-u_{i})=0$

$\lambda _{i}(l_{i}-x_{i})=0$

1.2 Покажите, что λi и μi не могут быть положительны одновременно

Предположим противное: 

$\lambda _{i}^{*}>0\quad \text{и}\quad \mu _{i}^{*}>0$

Из условия $\lambda_i^* (l_i - x_i^*) = 0$, так как $\lambda_i^* > 0$, с необходимостью следует, что:$l_{i}-x_{i}^{*}=0\implies x_{i}^{*}=l_{i}$

Из условия $\mu_i^* (x_i^* - u_i) = 0$, так как $\mu_i^* > 0$, аналогично следует, что: $x_{i}^{*}-u_{i}=0\implies x_{i}^{*}=u_{i}$

Получаем противоречие: 

$l_{i}=x_{i}^{*}=u_{i}\implies l_{i}=u_{i}$

Однако по условию задачи на коробочные ограничения границы не могут совпадать, интервал невырожденный: $l_i < u_i$

Вывод: переменные $\lambda _{i} и $\mu _{i}$ не могут быть одновременно строго положительными
