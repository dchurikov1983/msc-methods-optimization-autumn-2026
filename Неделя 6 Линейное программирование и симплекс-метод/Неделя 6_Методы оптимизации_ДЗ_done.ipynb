## Задача 1 (1.5 балла). Расписание смен операторов (ILP / MILP)

Кол-центр (или служба поддержки / диспетчерская) работает в течение горизонта $H=\{0,1,\ldots,|H|-1\}$ — например, часы суток или слоты по 30 минут. В каждый слот $h\in H$ нужно иметь на линии не менее $r_h$ операторов: величина $r_h$ задаётся прогнозом нагрузки (число звонков / тикетов) и нормативом обслуживания.

Операторов набирают **готовыми сменами** из каталога $K$. Смена $k\in K$ — это шаблон рабочего дня:

- покрывает подмножество слотов $H_k\subseteq H$ (например, «8:00–16:00 с обедом 12:00–13:00», тогда обеденный час в $H_k$ не входит);
- стоит $c_k$ (зарплата за смену, надбавка за ночь, аутсорс и тому подобное);
- может быть назначена не более $m_k$ раз ($m_k=1$, если смена уникальна для конкретного сотрудника; $m_k\ge 1$, если это тип смены, который можно выдать нескольким людям).

Решение — сколько раз взять каждую смену, чтобы **покрыть спрос по всем слотам** и **минимизировать суммарную стоимость**.

**Данные (вход модели).**

| Объект | Смысл |
|--------|--------|
| $H$ | множество временных слотов |
| $r_h\ge 0$ | минимальное число операторов в слоте $h$ |
| $K$ | каталог смен (типов смен) |
| $H_k\subseteq H$ | слоты, которые покрывает смена $k$ |
| $a_{hk}\in\{0,1\}$ | индикатор: $a_{hk}=1\Leftrightarrow h\in H_k$ |
| $c_k>0$ | стоимость одной копии смены $k$ |
| $m_k\in\mathbb{Z}_{++}$ | верхняя граница на число копий смены $k$ |

1. Переменные

Пусть переменная $x_k \ge 0$ означает, сколько раз берется смена типа $k \in K$ (число копий смены $k$). В рамках линейного программирования (линейной релаксации) переменные считаются непрерывными.

---

2. Математическая формулировка LP-задачи

Цель — минимизировать суммарную стоимость смен при соблюдении требований к минимальному количеству операторов в каждом слоте и ограничений на максимальное количество использований каждого типа смены:

$$ \min_{x} \sum_{k \in K} c_k x_k $$

При ограничениях:

1. Покрытие спроса в каждом временном слоте $h \in H$:
   $$ \sum_{k \in K} a_{hk} x_k \ge r_h, \quad \forall h \in H $$
2. Максимальное число копий смены каждого типа $k \in K$:
   $$ x_k \le m_k, \quad \forall k \in K $$
3. Неотрицательность переменных:
   $$ x_k \ge 0, \quad \forall k \in K $$

---

3. Компактная матричная запись

Пусть $n = |K|$ (количество типов смен), а $m = |H|$ (количество временных слотов). Введем векторы и матрицы:
* $x = (x_1, x_2, \dots, x_n)^\top \in \mathbb{R}^n$ — вектор переменных решения.
* $c = (c_1, c_2, \dots, c_n)^\top \in \mathbb{R}^n$ — вектор стоимостей смен.
* $r = (r_1, r_2, \dots, r_m)^\top \in \mathbb{R}^m$ — вектор минимального спроса по слотам.
* $m = (m_1, m_2, \dots, m_n)^\top \in \mathbb{R}^n$ — вектор верхних границ для смен.
* $A = [a_{hk}] \in \mathbb{R}^{m \times n}$ — матрица покрытия слотов сменами.
* $I_n \in \mathbb{R}^{n \times n}$, $I_m \in \mathbb{R}^{m \times m}$ — единичные матрицы.

В матрично-блочном виде задача записывается как:

$$ \min_{x} c^\top x $$

при ограничениях:
$$ \begin{pmatrix} A \\ -I_n \end{pmatrix} x \ge \begin{pmatrix} r \\ -m \end{pmatrix} $$
$$ x \ge 0 $$

---

4. Сведение блочной формы к стандартной

Стандартная форма задачи линейного программирования имеет вид:
$$ \min c_{\text{ст}}^\top x_{\text{ст}} \quad \text{при} \quad A_{\text{ст}} x_{\text{ст}} = b_{\text{ст}}, \quad x_{\text{ст}} \ge 0 $$

Для перехода от неравенств к равенствам введем неотрицательные дополнительные (слабые) переменные (slack variables):
1. Вектор $s = (s_1, \dots, s_m)^\top \ge 0$ для ограничений по слотам: $A x - s = r$.
2. Вектор $u = (u_1, \dots, u_n)^\top \ge 0$ для верхних границ смен: $x + u = m$.

Элементы стандартной формы собираются в следующие блоки:

* Расширенный вектор переменных:

  $$ x_{\text{ст}} = \begin{pmatrix} x \\ s \\ u \end{pmatrix} \in \mathbb{R}^{2n + m}, \quad x_{\text{ст}} \ge 0 $$

* Расширенный вектор стоимостей:

  $$ c_{\text{ст}} = \begin{pmatrix} c \\ 0_m \\ 0_n \end{pmatrix} \in \mathbb{R}^{2n + m} $$

* Составная матрица ограничений $A_{\text{ст}}$ и вектор правой части $b_{\text{ст}}$:

  $$ \begin{pmatrix} A & -I_m & 0_{m \times n} \\ I_n & 0_{n \times m} & I_n \end{pmatrix} \begin{pmatrix} x \\ s \\ u \end{pmatrix} = \begin{pmatrix} r \\ m \end{pmatrix} $$

  Где $A_{\text{ст}} = \begin{pmatrix} A & -I_m & 0_{m \times n} \\ I_n & 0_{n \times m} & I_n \end{pmatrix}$ имеет размерность $(m+n) \times (2n+m)$, а $b_{\text{ст}} = \begin{pmatrix} r \\ m \end{pmatrix}$ имеет размерность $(m+n)$.


Полное обоснование корректности математической модели

1. Почему любая допустимая точка модели соответствует осмысленному расписанию смен

Каждый вектор $x = (x_1, \dots, x_n)^\top$, удовлетворяющий ограничениям модели, гарантирует физическую реализуемость и бизнес-осмысленность расписания:
* Физическая реализуемость: Ограничение $x_k \ge 0$ исключает математический артефакт в виде «отрицательного» количества сотрудников на смене.
* Гарантия качества обслуживания (SL): Для каждого получасового слота $h \in H$ левая часть неравенства $\sum_{k \in K} a_{hk} x_k$ строго подсчитывает сумму всех работающих операторов (так как индикатор $a_{hk} = 1$ выделяет именно те смены, которые активны в этот момент). Выполнение условия $\ge r_h$ гарантирует, что на линии в любой момент времени $h$ будет достаточно людей для обработки входящего потока звонков/тикетов согласно нормативу. Просадки по сервису исключены.
* Соблюдение ограничений на ресурсы: Условие $x_k \le m_k$ гарантирует, что расписание не потребует вывода 10 человек на уникальную смену ($m_k=1$) или привлечения большего числа сотрудников, чем фактически числится в штате компании или доступно у аутсорс-провайдера.

---

2. Почему любое допустимое расписание из словесной постановки кодируется допустимой точкой

Пусть у менеджера кол-центра есть некоторое «бумажное» расписание, составленное вручную и удовлетворяющее всем словесным требованиям. Докажем, что его можно однозначно представить в виде допустимой точки нашей LP-модели:
* Мы можем просто посчитать, сколько раз в этом расписании используется каждый шаблон дня $k \in K$, и записать эти числа в компоненты вектора $x_k$. 
* Поскольку в реальном мире число людей на смене не может быть отрицательным, условие $x_k \ge 0$ выполнится автоматически.
* Так как «бумажное» расписание по условию является корректным, оно обязано покрывать прогнозируемый спрос. Значит, в каждый получас $h$ количество находящихся на рабочем месте операторов будет не меньше $r_h$. Математически это в точности означает, что вектор $x$ удовлетворит системе неравенств $\sum_{k \in K} a_{hk} x_k \ge r_h$.
* Аналогично, если расписание не нарушает ограничения на численность штата или условия уникальности сотрудников, то для каждого типа смены будет справедливо $x_k \le m_k$. 

Таким образом, ни одно валидное расписание не будет отсечено нашей системой ограничений.

---

3. Почему целевая функция совпадает со стоимостью выбранных смен

Целевая функция имеет вид линейной формы: $f(x) = \sum_{k \in K} c_k x_k$.
* Из словесной постановки нам известно, что параметр $c_k > 0$ — это фиксированная стоимость (зарплата за смену, надбавки, фиксированная ставка аутсорса) **одной копии** смены типа $k$.
* Переменная $x_k$ — это **количество выбранных копий** данной смены.
* В силу линейности, общие затраты на оплату труда складываются из затрат на каждый тип смен независимо друг от друга. Произведение стоимости одной смены на их количество ($c_k \cdot x_k$) с последующим суммированием по всему каталогу $K$ в точности по определению совпадает с суммарным фондом оплаты труда (ФОТ) кол-центра за рассматриваемый горизонт планирования $H$.

---

Перечисли принятые упрощения и то, как они ограничивают применимость модели к реальному кол-центру

Построенная модель является **линейной релаксацией** реальной задачи. При её составлении был допущен ряд серьезных упрощений, которые накладывают ограничения на применение модели в промышленной эксплуатации (production):

1. Непрерывность переменных ($x_k \in \mathbb{R}$)
   * Суть упрощения: Модель разрешает выводить дробное число сотрудников (например, $x_1 = 2.45$).  
   * Ограничение: На практике невозможно вывести на смену «половину» человека. Для малых кол-центров простое округление дробного решения до целого вверх (overstaffing) ведет к сильному завышению бюджета, а округление вниз — к просадке SL. Задача требует применения целочисленного программирования (ILP).
2. Абсолютная детерминированность спроса ($r_h$)
   * Суть упрощения: Минимально необходимое число операторов $r_h$ считается константой, известной заранее со 100% точностью.  
   * Ограничение: Поток звонков — это случайный процесс (обычно Пуассоновский). В реальности нагрузка колеблется. Если в слот $h$ придет неожиданный пиковый трафик, жестко зафиксированный норматив $r_h$ не спасет от огромной очереди, так как модель не учитывает стохастическую природу звонков и случайный фактор невыхода сотрудников на работу (absenteeism).
3. Статичность параметров смены ($a_{hk}$) 
   * Суть упрощения: Шаблон работы сотрудника внутри смены ($a_{hk}=1$ или $0$) зафиксирован намертво. Обеды и перерывы привязаны к строго определенным слотам.  
   * Ограничение: Реальные системы планирования (WFM — Workforce Management) должны гибко «двигать» обеды и технологические перерывы внутри смены сотрудника в зависимости от оперативной обстановки на линии, чтобы сглаживать пики нагрузки. Данная LP-модель этого делать не умеет, из-за чего каталог смен $K$ приходится искусственно раздувать до гигантских размеров (создавая отдельные типы смен на каждый возможный вариант смещения обеда на 30 минут).
4. Однородность (гомогенность) операторов
   * Суть упрощения: Считается, что любой оператор может обрабатывать любой поступивший тикет или звонок (все операторы обладают одинаковыми навыками/скиллами).  
   * Ограничение: В современных мультиканальных кол-центрах существует маршрутизация по навыкам (Skill-Based Routing). Например, одни операторы говорят на английском языке, другие консультируют по сложным техническим вопросам, третьи — по бухгалтерии. Данная модель может составить расписание, где общее число людей на линии формально правильное ($\ge r_h$), но в слоте окажутся только англоязычные сотрудники, а звонки на русском языке некому будет принять. Для учета этого фактора модель необходимо усложнять, вводя многопродуктовые потоки (индексы по навыкам).


# Здесь приведи интерпретацию найденного решения: какие смены взяты, сколько копий, хватает ли операторов в каждом слоте, где есть запас покрытия

• Смена 0: 4 копии (достигнут нижний порог для Слота 0)

• Смена 1: 5 копий (выбран весь доступный лимит штата m[1] = 5)

• Смена 2: 3 копии

• Смена 3: 5 копий (выбран весь доступный лимит штата m[3] = 5)

• Смена 4: 0 копий (модель полностью отказалась от этой смены, так как Смена 3 дешевле и полностью закрыла её слоты)

Операторов хватает в каждом слоте, причем в двух из них образовался избыток.

Запас покрытия (избыток операторов) образовался в двух временных интервалах:

• Слот 1: Выведено 9 операторов при спросе 7 (запас +2 оператора)

• Слот 5: Выведено 5 операторов при спросе 3 (запас +2 оператора)


````python
#!pip3 install pulp
# Пример стека: cvxpy + HiGHS / SCIP / CBC, либо PuLP + CBC
# import cvxpy as cp
import numpy as np
import pulp

# TODO: маленький экземпляр расписания смен:
#   H — слоты, r[h] — спрос, K — смены,
#   A[h,k] = 1 если смена k покрывает слот h,
#   c[k], m[k] — стоимость и верхняя граница.

H = list(range(6))

# Число операторов на смене
r = {0: 4, 1: 7, 2: 8, 3: 8, 4: 5, 5: 3}   

# Минимальный спрос на операторов по слотам                                             
m = {0: 5, 1: 5, 2: 4, 3: 5, 4: 4}         

# Каталог из 5 типов смен (переменных решения)
K = list(range(5))

# Стоимость смен (длинные или ночные смены обычно дороже)
c = {0: 100, 1: 120, 2: 150, 3: 90, 4: 110}

# Верхние границы на число копий смен (ограниченность штата)
#m = {0: 5, 1: 4, 2: 3, 3: 5, 4: 4}

# Матрица покрытия A[h, k] размера (6 слотов x 5 смен)
# 1 - если смена k покрывает слот h, 0 - иначе
A = np.array([
    [1, 0, 0, 0, 0],  # Слот 0: только смена 0
    [1, 1, 0, 0, 0],  # Слот 1: смены 0 и 1
    [0, 1, 1, 0, 0],  # Слот 2: смены 1 и 2
    [0, 0, 1, 1, 0],  # Слот 3: смены 2 и 3
    [0, 0, 0, 1, 1],  # Слот 4: смены 3 и 4
    [0, 0, 0, 1, 1]   # Слот 5: смены 3 и 4
])

milp_prob = pulp.LpProblem("CallCenter_MILP", pulp.LpMinimize)

# Ограничение целочисленности (cat='Integer') и верхние границы m[k]
x_milp = {
    k: pulp.LpVariable(f"x_{k}", lowBound=0, upBound=m[k], cat="Integer")
    for k in K
}

# Целевая функция
milp_prob += pulp.lpSum(c[k] * x_milp[k] for k in K)

# Ограничения покрытия спроса
for h in H:
    milp_prob += (
        pulp.lpSum(A[h, k] * x_milp[k] for k in K) >= r[h],
        f"Demand_Slot_{h}",
    )

# Решение
milp_prob.solve(pulp.PULP_CBC_CMD(msg=False))
milp_obj = pulp.value(milp_prob.objective)

lp_prob = pulp.LpProblem("CallCenter_LP", pulp.LpMinimize)

# Переменные теперь непрерывные (cat='Continuous')
x_lp = {
    k: pulp.LpVariable(f"x_{k}", lowBound=0, upBound=m[k], cat="Continuous")
    for k in K
}

# Целевая функция
lp_prob += pulp.lpSum(c[k] * x_lp[k] for k in K)

# Ограничения покрытия спроса
for h in H:
    lp_prob += (
        pulp.lpSum(A[h, k] * x_lp[k] for k in K) >= r[h],
        f"Demand_Slot_{h}",
    )

# Решение
lp_prob.solve(pulp.PULP_CBC_CMD(msg=False))
lp_relax_obj = pulp.value(lp_prob.objective)




#abs_gap = milp_obj - lp_relax_obj

#gap = None  # |milp - lp| / max(1, |milp|) или в абсолютных единицах
gap = abs(milp_obj - lp_relax_obj) / max(1.0, abs(milp_obj))

print("MILP:", milp_obj)
print("LP relaxation:", lp_relax_obj)
print("gap:", gap)
````

Значения milp_obj и lp_relax_obj совпали (обе целевые функции равны 1900.0, а gap = 0.0), это означает, что для данного небольшого примера непрерывная LP-релаксация сразу нашла строго целочисленное решение.

### 2.1. Экземпляр и настройки

| Экземпляр | Источник | #var | #int | #cons | time limit | threads |
|-----------|----------|------|------|-------|------------|---------|
| `nursesched-medium04` | MIPLIB 2017 | ~29667 | ~29613 bin | ~8668 | 120 | 1 |

````python
from pathlib import Path
import urllib.request
import gzip
import shutil
import time

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)
URL = "https://miplib.zib.de/WebData/instances/nursesched-medium04.mps.gz"
GZ_PATH = DATA_DIR / "nursesched-medium04.mps.gz"
MPS_PATH = DATA_DIR / "nursesched-medium04.mps"
TIME_LIMIT = 120
THREADS = 1
KNOWN_OPT = 237  # MIPLIB / INRC medium04

if not MPS_PATH.exists():
    urllib.request.urlretrieve(URL, GZ_PATH)
    with gzip.open(GZ_PATH, "rb") as f_in, open(MPS_PATH, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)

print("MPS:", MPS_PATH.resolve(), "size=", MPS_PATH.stat().st_size)


# --- вариант A: HiGHS ---
import highspy
h = highspy.Highs()
h.readModel(str(MPS_PATH))
h.setOptionValue("time_limit", TIME_LIMIT)
h.setOptionValue("threads", THREADS)
h.run()
print(h.getModelStatus(), h.getObjectiveValue())

# --- вариант B: LP-релаксация (HiGHS relax) ---
import highspy

h_relax = highspy.Highs()
h_relax.readModel(str(MPS_PATH))

# Превращаем все переменные в непрерывные (снимаем целочисленные ограничения)
num_cols = h_relax.getNumCol()
for i in range(num_cols):
    # Изменяем тип переменной на CONTINUOUS (непрерывная)
    h_relax.changeColIntegrality(i, highspy.HighsVarType.kContinuous)

# Устанавливаем параметры для LP
h_relax.setOptionValue("time_limit", TIME_LIMIT)
h_relax.setOptionValue("threads", THREADS)

# Запускаем LP-релаксации (обычно решается симплекс-методом или методом внутренних точек)
print("\nЗапуск LP-релаксации...")
t_start = time.time()
h_relax.run()
lp_time = time.time() - t_start

print(f"Статус LP: {h_relax.getModelStatus()}")
print(f"Нижняя граница (LP objective): {h_relax.getObjectiveValue():.4f}")
print(f"Время решения LP: {lp_time:.2f} с")


# --- Сравнение результатов (Метрики бенчмарка) ---
print("\n=== Результаты бенчмарка ===")
print(f"Известный оптимум (MIP): {KNOWN_OPT}")
if 'h' in locals() and h.getModelStatus() == highspy.HighsModelStatus.kOptimal:
    mip_obj = h.getObjectiveValue()
    print(f"Найденный MIP оптимум: {mip_obj}")
    gap = (mip_obj - KNOWN_OPT) / KNOWN_OPT * 100
    print(f"MIP Gap относительно известного оптимума: {gap:.2f}%")

lp_obj = h_relax.getObjectiveValue()
lp_gap = (KNOWN_OPT - lp_obj) / KNOWN_OPT * 100
print(f"LP-отклонение (Integrality Gap): {lp_gap:.2f}%")
````

````python
# --- TODO: прогоны и сбор метрик ---
results = []

# --- Прогон Варианта A: Полноценный MILP (HiGHS) ---
print("Запуск Варианта A (MILP)...")
h_mip = highspy.Highs()
h_mip.readModel(str(MPS_PATH))
h_mip.setOptionValue("time_limit", float(TIME_LIMIT))
h_mip.setOptionValue("threads", int(THREADS))

t_start_mip = time.time()
h_mip.run()
time_mip = time.time() - t_start_mip

# Сбор статистики через объект Info
info_mip = h_mip.getInfo()
status_mip = h_mip.getModelStatus()

results.append({
    "instance": MPS_PATH.stem,
    "solver": "HiGHS (MILP)",
    "status": str(status_mip).split(".")[-1],  # Красивое имя статуса (например, kOptimal или kTimeLimit)
    "best_feasible": h_mip.getObjectiveValue(),
    "best_bound": info_mip.mip_dual_bound if hasattr(info_mip, 'mip_dual_bound') else None,
    "gap": info_mip.mip_gap if hasattr(info_mip, 'mip_gap') else None,
    "time_sec": time_mip,
    "nodes": info_mip.mip_node_count if hasattr(info_mip, 'mip_node_count') else 0,
})


# --- Прогон Варианта B: LP-релаксация (HiGHS relax) ---
print("Запуск Варианта B (LP-релаксация)...")
h_relax = highspy.Highs()
h_relax.readModel(str(MPS_PATH))

# Снимаем целочисленные ограничения со всех переменных
num_cols = h_relax.getNumCol()
for i in range(num_cols):
    h_relax.changeColIntegrality(i, highspy.HighsVarType.kContinuous)

h_relax.setOptionValue("time_limit", float(TIME_LIMIT))
h_relax.setOptionValue("threads", int(THREADS))

t_start_lp = time.time()
h_relax.run()
time_lp = time.time() - t_start_lp

info_lp = h_relax.getInfo()
status_lp = h_relax.getModelStatus()
lp_obj = h_relax.getObjectiveValue()

# Для LP-релаксации понятие зазора (gap) и узлов (nodes) неприменимо,
# но мы можем рассчитать относительное LP-отклонение (Integrality Gap) вручную от известного оптимума.
integrality_gap = (KNOWN_OPT - lp_obj) / KNOWN_OPT if KNOWN_OPT > 0 else 0

results.append({
    "instance": MPS_PATH.stem,
    "solver": "HiGHS (LP-Relax)",
    "status": str(status_lp).split(".")[-1],
    "best_feasible": lp_obj,
    "best_bound": lp_obj,  # В LP целевая функция является одновременно и точной границей
    "gap": integrality_gap,  # Сохраняем вычисленный Integrality Gap
    "time_sec": time_lp,
    "nodes": 0,  # Узлов ветвления в LP нет
})


# --- Создание и вывод таблицы результатов ---
import pandas as pd

df = pd.DataFrame(results)
# Форматируем вывод для удобства чтения
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
print("\n=== Итоговая таблица бенчмарка ===")
print(df)
````

### 2.3. Сравнение

*Кто быстрее на релаксации? Кто устойчивее на MILP? Как это стыкуется с лекцией?*

Ответ:
Быстрее всех на релаксации открытый солвер CLP (или связка HiGHS / COPT при использовании симплекс-метода).

Как это стыкуется с лекцией: На слайде 10 («Солверы: лицензии и цены») и слайде 11 («Независимые бенчмарки») четко показано разделение. Для чистых ЛП-задач открытые солверы, такие как HiGHS и CLP, работают превосходно. В тестах Миттельманна на ЛП открытый солвер CLP отстает от коммерческого лидера COPT всего в 27 раз, а HiGHS — в 17 раз. Это очень хороший результат для Open Source, что мы и увидели на практике: HiGHS (LP-Relax) справился с огромной матрицей (почти 30 000 переменных) всего за 1.04 секунды.
