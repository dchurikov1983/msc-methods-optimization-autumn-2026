 1. Выбираем произвольные точки
 Пусть две точки $(\mathbf{x}_1, t_1)$ и $(\mathbf{x}_2, t_2)$ принадлежат множеству $\mathcal{L}^{n+1}$. По определению этого множества выполняются два неравенства:

a. $\Vert{}\mathbf{x}_1\Vert{}_2 \leqslant t_1$

b. $\Vert{}\mathbf{x}_2\Vert{}_2 \leqslant t_2$


2. Рассмотрим их выпуклую комбинацию
коэффициент $\lambda \in [0, 1]$

$w=\lambda u+(1-\lambda )v=\big(\lambda x_{1}+(1-\lambda )x_{2},\,\lambda t_{1}+(1-\lambda )t_{2}\big)$


3. При условии принадлежности точки $w$ конусу $L^{n+1}$ 

Необходимо доказать:

$\|{}\lambda x_{1}+(1-\lambda )x_{2}\|{}_{2}\leqslant \lambda t_{1}+(1-\lambda )t_{2}$


4. Применим свойство неравенства треугольника для евклидовой нормы:

$\Vert{}\lambda x_1 + (1-\lambda)x_2\Vert{}_2 \leqslant \Vert{}\lambda x_1\Vert{}_2 + \Vert{}(1-\lambda)x_2\Vert{}_2$


5. Так как $\lambda \geqslant 0$ и $(1-\lambda) \geqslant 0$, мы можем вынести эти скаляры за знак нормы:

$\Vert{}\lambda x_1\Vert{}_2 + \Vert{}(1-\lambda)x_2\Vert{}_2 = \lambda \Vert{}x_1\Vert{}_2 + (1-\lambda) \Vert{}x_2\Vert{}_2$


6. Используя  $\Vert{}x_1\Vert{}_2 \leqslant t_1$ и $\Vert{}x_2\Vert{}_2 \leqslant t_2$, запишем :

$\lambda \Vert{}x_1\Vert{}_2 + (1-\lambda) \Vert{}x_2\Vert{}_2 \leqslant \lambda t_1 + (1-\lambda) t_2$


7. получаем:
$\Vert{}\lambda x_1 + (1-\lambda)x_2\Vert{}_2 \leqslant \lambda t_1 + (1-\lambda) t_2$


8.  $(x_\lambda, t_\lambda) \in L^{n+1}$. Множество выпукло, что и требовалось доказать.

3.1 Для каждой функции определите является ли она выпуклой / вогнутой / ни то ни другое
на указанной области. 
Функция Хьюбера на R:
h(r) =
{12r2, |r|⩽1,
|r|−12, |r|> 1

Функция называется выпуклой (вниз) на промежутке, если ее первая производная является неубывающей функцией на этом промежутке. Найдем производную функции h(r):

При $(\vert{}r\vert{} \le 1)$ $h'(r) = \left(\frac{1}{2}r^2\right)' = r$
	
При $(\vert{}r\vert{} > 1)$ $(h'(r) = \left(\vert{}r\vert{} - \frac{1}{2}\right)' = \text{sign}(r))$

, где $\text{sign}(r) = 1$ при $(r > 1)$ и $(-1)$ при $r < -1$


=> 
$h^{\prime }(r)=\begin{cases}-1,&r<-1\\ r,&-1\le r\le 1\\ 1,&r>1\end{cases}$

Так как производная $h'(r)$ является непрерывной и неубывающей на всей области $\mathbb{R}$, исходная функция h(r) является выпуклой.

Какая интерпретация этой функции?

 Её главная интерпретация — это устойчивая (робастная) замена квадратичной ошибки

В каких случаях её уместно использовать в качестве целевой функции?

Функция Хьюбера (Huber loss) широко применяется в машинном обучении, статистике и оптимизации как функция потерь. Она сочетает в себе лучшие свойства двух классических метрик: среднеквадратичной ошибки (MSE / L2-потери) и средней абсолютной ошибки (MAE / L1-потери)

3.2 Проверьте выпуклость по совокупности аргументов:
f (x,Y) = x⊤Y−1xна x∈Rn, Y∈Sn++

1. Определение надграфика (epigraph):

Надграфик функции f — это множество точек: 

$\mathrm{epi}\>f=\{(\mathbf{x},\mathbf{Y},t)\mid \mathbf{x}^{\top }\mathbf{Y}^{-1}\mathbf{x}\le t,\ \mathbf{Y}\succ 0\}$

2. Применение дополнения Шура 

для симметричной положительно определенной матрицы $\mathbf{Y} \succ 0$ матричное неравенство:

$t-\mathbf{x}^{\top }\mathbf{Y}^{-1}\mathbf{x}\ge 0$

ыполняется тогда и только тогда, когда

$\left[\begin{matrix}\mathbf{Y}&\mathbf{x}\\ \mathbf{x}^{\top }&t\end{matrix}\right]\succeq 0$

Вывод:

Множество, задаваемое линейным матричным неравенством, всегда является выпуклым. Поскольку надграфик $\operatorname{epi} f$ выпуклый, сама функция $f(\mathbf{x}, \mathbf{Y})$ является выпуклой по совокупности аргументов. 

3.2 Проверить на выпуклость/вогнутость множество:
Геометрическое среднее на Rn++:
g(x) = (x1x2 ···xn)1/n

Согласно материалам лонгрида:
1. Функция вогнута, если функция $-g(\mathbf{x})$ выпукла.
2. Дифференцируемая функция выпукла тогда и только тогда, когда её матрица вторых производных (Гессиан) является положительно полуопределенной $(H_{-g}(\mathbf{x}) \succeq 0)$, что эквивалентно отрицательной полуопределенности Гессиана самой функции $g(\mathbf{x})$, то есть $H_g(\mathbf{x}) \preceq 0$.


1. Находим частные производных первого порядка

$\frac{\partial g}{\partial x_{i}}=\frac{1}{n}(x_{1}x_{2}\cdots x_{n})^{\frac{1}{n}-1}\cdot \prod _{k\ne i}x_{k}=\frac{1}{n}\frac{g(\mathbf{x})}{x_{i}}$

2. Находим частные производные второго порядка

i = j

$\frac{\partial ^{2}g}{\partial x_{i}^{2}}=\frac{1}{n}\frac{\partial }{\partial x_{i}}\left(\frac{g(\mathbf{x})}{x_{i}}\right)=\frac{1}{n}\left(\frac{\frac{1}{n}\frac{g(\mathbf{x})}{x_{i}}\cdot x_{i}-g(\mathbf{x})}{x_{i}^{2}}\right)=\frac{1-n}{n^{2}}\frac{g(\mathbf{x})}{x_{i}^{2}}$

i!=j

$\frac{\partial ^{2}g}{\partial x_{i}\partial x_{j}}=\frac{1}{nx_{i}}\frac{\partial g}{\partial x_{j}}=\frac{1}{n^{2}}\frac{g(\mathbf{x})}{x_{i}x_{j}}$


3. Проверка знака квадратичной формы

$\mathbf{y}^{\top }H_{g}(\mathbf{x})\mathbf{y}=\sum _{i=1}^{n}\sum _{j=1}^{n}\frac{\partial ^{2}g}{\partial x_{i}\partial x_{j}}y_{i}y_{j}$

$\mathbf{y}^{\top }H_{g}(\mathbf{x})\mathbf{y}=\sum _{i\ne j}\frac{g(\mathbf{x})}{n^{2}x_{i}x_{j}}y_{i}y_{j}+\sum _{i=1}^{n}\frac{(1-n)g(\mathbf{x})}{n^{2}x_{i}^{2}}y_{i}^{2}$

$\mathbf{y}^{\top }H_{g}(\mathbf{x})\mathbf{y}=\frac{g(\mathbf{x})}{n^{2}}\left[\sum _{i=1}^{n}\sum _{j=1}^{n}\frac{y_{i}y_{j}}{x_{i}x_{j}}-n\sum _{i=1}^{n}\frac{y_{i}^{2}}{x_{i}^{2}}\right]$

$\mathbf{y}^{\top }H_{g}(\mathbf{x})\mathbf{y}=\frac{g(\mathbf{x})}{n^{2}}\left[\left(\sum _{i=1}^{n}\frac{y_{i}}{x_{i}}\right)^{2}-n\sum _{i=1}^{n}\frac{y_{i}^{2}}{x_{i}^{2}}\right]$



4. Применим неравенства Коши-Буняковского-Шварца

$z_i = \frac{y_i}{x_i}$

$\left(\sum _{i=1}^{n}\frac{y_{i}}{x_{i}}\right)^{2}=\left(\sum _{i=1}^{n}z_{i}\cdot 1\right)^{2}$

По неравенству Коши-Буняковского-Шварца:
$\left(\sum _{i=1}^{n}z_{i}\cdot 1\right)^{2}\leqslant \left(\sum _{i=1}^{n}z_{i}^{2}\right)\left(\sum _{i=1}^{n}1^{2}\right)=\left(\sum _{i=1}^{n}\frac{y_{i}^{2}}{x_{i}^{2}}\right)\cdot n$

Следовательно:
$\left(\sum _{i=1}^{n}\frac{y_{i}}{x_{i}}\right)^{2}-n\sum _{i=1}^{n}\frac{y_{i}^{2}}{x_{i}^{2}}\leqslant 0$


Вывод

Так как на множестве $\mathbb{R}_{++}^{n}$ значения $g(\mathbf{x}) > 0$, вся квадратичная форма $\mathbf{y}^\top H_g(\mathbf{x}) \mathbf{y} \le 0$ для любых $\mathbf{y}$. Это означает, что матрица Гессиана отрицательно полуопределена ($H_g(\mathbf{x}) \preceq 0$), а значит, по критерию 2-го порядка функция геометрического среднего является вогнутой. Что и требовалось доказать.
