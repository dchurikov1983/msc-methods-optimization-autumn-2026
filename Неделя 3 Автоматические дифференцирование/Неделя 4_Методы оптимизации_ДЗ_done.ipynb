1.1 Докажите выпуклость конуса второго порядка (Lorentz cone) Ln+1 = {(x,t) ∈Rn ×R:
∥x∥2 ⩽t}

Из Лекции:
Множество S⊆ Rn выпукло, если вместе с любыми двумя точками оно содержит весь отрезок между
ними:
∀x,y ∈ S,λ∈ [0,1] ⇒ λx + (1 − λ)y ∈ S

 1. Выбираем произвольные точки
 Пусть две точки $(\mathbf{x}_1, t_1)$ и $(\mathbf{x}_2, t_2)$ принадлежат множеству $\mathcal{L}^{n+1}$. По определению этого множества выполняются два неравенства:

a. $\Vert{}\mathbf{x}_1\Vert{}_2 \leqslant t_1$

b. $\Vert{}\mathbf{x}_2\Vert{}_2 \leqslant t_2$


2. Рассмотрим их выпуклую комбинацию
коэффициент $\lambda \in [0, 1]$

$w=\lambda u+(1-\lambda )v=\big(\lambda x_{1}+(1-\lambda )x_{2},\,\lambda t_{1}+(1-\lambda )t_{2}\big)$


3. При условии принадлежности точки $w$ конусу $L^{n+1}$ 

Необходимо доказать:

$\|{}\lambda x_{1}+(1-\lambda )x_{2}\|{}_{2}\leqslant \lambda t_{1}+(1-\lambda )t_{2}$


4. Применим свойство неравенства треугольника для евклидовой нормы:

$\Vert{}\lambda x_1 + (1-\lambda)x_2\Vert{}_2 \leqslant \Vert{}\lambda x_1\Vert{}_2 + \Vert{}(1-\lambda)x_2\Vert{}_2$


5. Так как $\lambda \geqslant 0$ и $(1-\lambda) \geqslant 0$, мы можем вынести эти скаляры за знак нормы:

$\Vert{}\lambda x_1\Vert{}_2 + \Vert{}(1-\lambda)x_2\Vert{}_2 = \lambda \Vert{}x_1\Vert{}_2 + (1-\lambda) \Vert{}x_2\Vert{}_2$


6. Используя  $\Vert{}x_1\Vert{}_2 \leqslant t_1$ и $\Vert{}x_2\Vert{}_2 \leqslant t_2$, запишем :

$\lambda \Vert{}x_1\Vert{}_2 + (1-\lambda) \Vert{}x_2\Vert{}_2 \leqslant \lambda t_1 + (1-\lambda) t_2$


7. получаем:
$\Vert{}\lambda x_1 + (1-\lambda)x_2\Vert{}_2 \leqslant \lambda t_1 + (1-\lambda) t_2$


8.  $(x_\lambda, t_\lambda) \in L^{n+1}$. Множество выпукло, что и требовалось доказать.
